```python
# Cell 1 — point the notebook at the remote server instead of a local folder
import mlflow

mlflow.set_tracking_uri("http://localhost:5000")
print("MLflow client version:", mlflow.__version__)
print("Tracking URI:", mlflow.get_tracking_uri())
```

```python
# Cell 2 — create (or reuse) an experiment and log one run
mlflow.set_experiment("00_smoke_test")

with mlflow.start_run(run_name="hello-mlflow") as run:
    # PARAMETER = a setting you chose (later: max_depth, threshold...)
    mlflow.log_param("purpose", "check the server works")

    # METRIC = a number you measured (later: pr_auc, wape...)
    mlflow.log_metric("answer", 42)

    # A metric logged over steps -> MLflow draws a line chart
    for step in range(5):
        mlflow.log_metric("fake_loss", 1 / (step + 1), step=step)

    # ARTIFACT = a file (later: the model, SHAP plots, model card)
    mlflow.log_text("If you can read this, artifact storage works.", "notes/hello.txt")

    print("Run ID:", run.info.run_id)
```
